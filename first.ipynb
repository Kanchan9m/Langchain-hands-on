```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate
from langchain_groq import ChatGroq
```

```python
#chat Template with system + user roles
sys_msg = SystemMessagePromptTemplate.from_template("You are a concise Assistant; give short factual answers.")
user_msg = HumanMessagePromptTemplate.from_template("{question_text}")
```

```python
chat_template = ChatPromptTemplate.from_messages([sys_msg, user_msg])
```

```python
groq_chat = ChatGroq(model="llama-3.1-8b-instant", temperature=0.0)
```

```python
#string parser and pipeline
# it will parse the raw output from model into a plain string
text_only_parser = StrOutputParser()
# | operator is used to chain the prompt, model and parser together into a single pipeline 
string_pipeline = chat_template | groq_chat | text_only_parser
```

```python
result = string_pipeline.invoke({"question_text": "What is Machine Learning?"})
print("Raw result: ", result)
```
